```python
# Make sure matches are in chronological order
df2 = df2.sort_values("date").reset_index(drop=True)

# Starting Elo rating for every team
INITIAL_ELO = 1500
K_FACTOR = 20

# Dictionary to store current ratings
elo_ratings = {}

# Lists to store pre-match and post-match Elo values
home_elo_before = []
away_elo_before = []
home_elo_after = []
away_elo_after = []
```

```python
def expected_score(rating_a, rating_b):
    """
    Expected probability of Team A getting a result against Team B.
    """
    return 1 / (1 + 10 ** ((rating_b - rating_a) / 400))


def actual_scores(home_score, away_score):
    """
    Returns Elo result values:
    Win = 1
    Draw = 0.5
    Loss = 0
    """
    if home_score > away_score:
        return 1.0, 0.0
    elif home_score < away_score:
        return 0.0, 1.0
    else:
        return 0.5, 0.5
```

```python
for _, row in df2.iterrows():

    home = row["home_team"]
    away = row["away_team"]

    # Give unseen teams the default rating
    home_rating = elo_ratings.get(home, INITIAL_ELO)
    away_rating = elo_ratings.get(away, INITIAL_ELO)

    # Store ratings before the match
    home_elo_before.append(home_rating)
    away_elo_before.append(away_rating)

    # Expected outcome
    home_expected = expected_score(home_rating, away_rating)
    away_expected = expected_score(away_rating, home_rating)

    # Actual outcome
    home_actual, away_actual = actual_scores(
        row["home_score"],
        row["away_score"]
    )

    # Update Elo
    new_home_rating = home_rating + K_FACTOR * (
        home_actual - home_expected
    )

    new_away_rating = away_rating + K_FACTOR * (
        away_actual - away_expected
    )

    # Save updated ratings
    elo_ratings[home] = new_home_rating
    elo_ratings[away] = new_away_rating

    home_elo_after.append(new_home_rating)
    away_elo_after.append(new_away_rating)
```

```python
df2["home_elo_before"] = home_elo_before
df2["away_elo_before"] = away_elo_before
df2["home_elo_after"] = home_elo_after
df2["away_elo_after"] = away_elo_after
```

```python
HOME_ADVANTAGE = 100

def expected_score_with_home(
    home_rating,
    away_rating,
    neutral
):
    
    # Only apply home advantage for non-neutral matches
    adjusted_home_rating = (
        home_rating
        if neutral
        else home_rating + HOME_ADVANTAGE
    )
    
    home_expected = 1 / (
        1 + 10 ** (
            (away_rating - adjusted_home_rating) / 400
        )
    )
    
    away_expected = 1 - home_expected
    
    return home_expected, away_expected
```

```python
INITIAL_ELO = 1500
K_FACTOR = 20

elo_ratings_v2 = {}

home_elo_v2_before = []
away_elo_v2_before = []
home_expected_v2 = []

for _, row in df2.iterrows():
    
    home = row["home_team"]
    away = row["away_team"]
    
    home_rating = elo_ratings_v2.get(home, INITIAL_ELO)
    away_rating = elo_ratings_v2.get(away, INITIAL_ELO)
    
    home_elo_v2_before.append(home_rating)
    away_elo_v2_before.append(away_rating)
    
    # Expected result with home advantage
    home_expected, away_expected = expected_score_with_home(
        home_rating,
        away_rating,
        row["neutral"]
    )
    
    home_expected_v2.append(home_expected)
    
    # Actual result
    home_actual, away_actual = actual_scores(
        row["home_score"],
        row["away_score"]
    )
    
    # Elo updates
    new_home_rating = (
        home_rating
        + K_FACTOR * (home_actual - home_expected)
    )
    
    new_away_rating = (
        away_rating
        + K_FACTOR * (away_actual - away_expected)
    )
    
    elo_ratings_v2[home] = new_home_rating
    elo_ratings_v2[away] = new_away_rating
```

```python
df2["home_elo_v2_before"] = home_elo_v2_before
df2["away_elo_v2_before"] = away_elo_v2_before
df2["home_expected_v2"] = home_expected_v2
```

```python
def run_elo(matches, ratings=None, home_advantage=0, k_factor=20):
    
    if ratings is None:
        ratings = {}
    else:
        ratings = ratings.copy()

    predictions = []

    for _, row in matches.iterrows():

        home = row["home_team"]
        away = row["away_team"]

        home_rating = ratings.get(home, 1500)
        away_rating = ratings.get(away, 1500)

        adjusted_home = (
            home_rating
            if row["neutral"]
            else home_rating + home_advantage
        )

        home_expected = 1 / (
            1 + 10 ** (
                (away_rating - adjusted_home) / 400
            )
        )

        away_expected = 1 - home_expected

        predictions.append(home_expected)

        home_actual, away_actual = actual_scores(
            row["home_score"],
            row["away_score"]
        )

        ratings[home] = (
            home_rating
            + k_factor * (home_actual - home_expected)
        )

        ratings[away] = (
            away_rating
            + k_factor * (away_actual - away_expected)
        )

    return ratings, predictions
```

```python
def tournament_tier(tournament):

    t = tournament.lower().strip()

    # World Cup finals
    if t == "world cup":
        return 5

    # Major continental championships
    major_championships = [
        "african nations cup",
        "european championship",
        "asian cup",
        "copa america",
        "copa américa",
        "concacaf championship",
        "oceania nations cup"
    ]

    if t in major_championships:
        return 4

    # Qualifiers
    if (
        "qualifier" in t
        or " qual" in t
        or t.endswith(" q")
    ):
        return 3

    # Friendlies
    if "friendly" in t:
        return 1

    # Everything else:
    # Nations League, COSAFA, CECAFA, Gulf Cup, etc.
    return 2
```

```python
TOURNAMENT_WEIGHTS = {
    1: 0.75,
    2: 1.00,
    3: 1.25,
    4: 1.50,
    5: 2.00
}
```

```python
def run_elo_weighted(
    matches,
    ratings=None,
    home_advantage=80,
    k_factor=20,
    tournament_weights=TOURNAMENT_WEIGHTS
):

    if ratings is None:
        ratings = {}
    else:
        ratings = ratings.copy()

    predictions = []

    for _, row in matches.iterrows():

        home = row["home_team"]
        away = row["away_team"]

        home_rating = ratings.get(home, 1500)
        away_rating = ratings.get(away, 1500)

        # Home advantage
        adjusted_home = (
            home_rating
            if row["neutral"]
            else home_rating + home_advantage
        )

        # Expected result
        home_expected = 1 / (
            1 + 10 ** (
                (away_rating - adjusted_home) / 400
            )
        )

        away_expected = 1 - home_expected

        predictions.append(home_expected)

        # Actual result
        home_actual, away_actual = actual_scores(
            row["home_score"],
            row["away_score"]
        )

        # Tournament-adjusted K
        tier = row["tournament_tier"]

        match_k = (
            k_factor *
            tournament_weights[tier]
        )

        # Update ratings
        ratings[home] = (
            home_rating +
            match_k * (home_actual - home_expected)
        )

        ratings[away] = (
            away_rating +
            match_k * (away_actual - away_expected)
        )

    return ratings, predictions
```

```python
train_df = df2[df2["year"] < 2015].copy()

val_df = df2[
    (df2["year"] >= 2015) &
    (df2["year"] < 2022)
].copy()

test_df = df2[df2["year"] >= 2022].copy()
```

```python
def run_elo_with_history(
    matches,
    home_advantage=80,
    k_factor=20
):
    
    ratings = {}
    history = []

    for _, row in matches.iterrows():

        home = row["home_team"]
        away = row["away_team"]

        home_rating = ratings.get(home, 1500)
        away_rating = ratings.get(away, 1500)

        adjusted_home = (
            home_rating
            if row["neutral"]
            else home_rating + home_advantage
        )

        home_expected = 1 / (
            1 + 10 ** (
                (away_rating - adjusted_home) / 400
            )
        )

        away_expected = 1 - home_expected

        home_actual, away_actual = actual_scores(
            row["home_score"],
            row["away_score"]
        )

        new_home_rating = (
            home_rating +
            k_factor * (home_actual - home_expected)
        )

        new_away_rating = (
            away_rating +
            k_factor * (away_actual - away_expected)
        )

        ratings[home] = new_home_rating
        ratings[away] = new_away_rating

        # Save rating after every match
        history.append({
            "date": row["date"],
            "team": home,
            "elo": new_home_rating
        })

        history.append({
            "date": row["date"],
            "team": away,
            "elo": new_away_rating
        })

    return ratings, pd.DataFrame(history)
```

```python
continental_map = {
    "European Championship": "EURO",
    "Copa America": "COPA",
    "Copa América": "COPA",
    "African Nations Cup": "AFCON",
    "Asian Cup": "ASIAN_CUP",
    "CONCACAF Championship": "CONCACAF",
    "Oceania Nations Cup": "OFC"
}

continental = df2[
    df2["tournament"].isin(continental_map.keys())
].copy()

continental["competition"] = (
    continental["tournament"]
    .map(continental_map)
)

continental["powerhouse_home"] = (
    continental["home_team"]
    .replace({"West Germany": "Germany"})
)

continental["powerhouse_away"] = (
    continental["away_team"]
    .replace({"West Germany": "Germany"})
)
```

```python
# ============================================================
# CONTINENTAL CHAMPIONSHIP PERFORMANCE
# ============================================================

continental["home_points"] = 0
continental["away_points"] = 0

# Home wins
home_win = continental["home_score"] > continental["away_score"]
continental.loc[home_win, "home_points"] = 3

# Away wins
away_win = continental["away_score"] > continental["home_score"]
continental.loc[away_win, "away_points"] = 3

# Draws
draw = continental["home_score"] == continental["away_score"]
continental.loc[draw, ["home_points", "away_points"]] = 1


# ------------------------------------------------------------
# Convert every match into team-level rows
# ------------------------------------------------------------

continental_home = pd.DataFrame({
    "team": continental["powerhouse_home"],
    "competition": continental["competition"],
    "year": continental["year"],
    "GF": continental["home_score"],
    "GA": continental["away_score"],
    "points": continental["home_points"]
})

continental_away = pd.DataFrame({
    "team": continental["powerhouse_away"],
    "competition": continental["competition"],
    "year": continental["year"],
    "GF": continental["away_score"],
    "GA": continental["home_score"],
    "points": continental["away_points"]
})

continental_team_matches = pd.concat(
    [continental_home, continental_away],
    ignore_index=True
)

continental_team_matches["GD"] = (
    continental_team_matches["GF"]
    - continental_team_matches["GA"]
)

continental_team_matches["win"] = (
    continental_team_matches["points"] == 3
).astype(int)

continental_team_matches["draw"] = (
    continental_team_matches["points"] == 1
).astype(int)

continental_team_matches["loss"] = (
    continental_team_matches["points"] == 0
).astype(int)
```

```python
continental_performance = (
    continental_team_matches
    .groupby("team")
    .agg(
        matches=("team", "size"),
        wins=("win", "sum"),
        draws=("draw", "sum"),
        losses=("loss", "sum"),
        goals_for=("GF", "sum"),
        goals_against=("GA", "sum"),
        goal_difference=("GD", "sum"),
        points=("points", "sum")
    )
)

continental_performance["win_pct"] = (
    continental_performance["wins"]
    / continental_performance["matches"]
    * 100
)

continental_performance["points_per_game"] = (
    continental_performance["points"]
    / continental_performance["matches"]
)

continental_performance["goal_difference_per_game"] = (
    continental_performance["goal_difference"]
    / continental_performance["matches"]
)

continental_performance["goals_per_game"] = (
    continental_performance["goals_for"]
    / continental_performance["matches"]
)

continental_performance[
    continental_performance["matches"] >= 20
].sort_values(
    "points_per_game",
    ascending=False
).head(30)
```

```python
# Add each team's primary continental competition
team_competition = (
    continental_team_matches
    .groupby("team")["competition"]
    .agg(lambda x: x.value_counts().index[0])
)

continental_performance["competition"] = team_competition


# Only use teams with enough matches for a meaningful sample
continental_valid = continental_performance[
    continental_performance["matches"] >= 20
].copy()


# ---------------------------------------------------------
# Z-SCORE PERFORMANCE WITHIN EACH CONFEDERATION
# ---------------------------------------------------------

continental_valid["ppg_z"] = (
    continental_valid
    .groupby("competition")["points_per_game"]
    .transform(
        lambda x: (x - x.mean()) / x.std()
    )
)

continental_valid["gd_z"] = (
    continental_valid
    .groupby("competition")["goal_difference_per_game"]
    .transform(
        lambda x: (x - x.mean()) / x.std()
    )
)

continental_valid["win_z"] = (
    continental_valid
    .groupby("competition")["win_pct"]
    .transform(
        lambda x: (x - x.mean()) / x.std()
    )
)


# Composite continental dominance score
continental_valid["continental_dominance"] = (
    0.50 * continental_valid["ppg_z"]
    + 0.30 * continental_valid["gd_z"]
    + 0.20 * continental_valid["win_z"]
)


continental_valid[
    [
        "competition",
        "matches",
        "win_pct",
        "points_per_game",
        "goal_difference_per_game",
        "continental_dominance"
    ]
].sort_values(
    "continental_dominance",
    ascending=False
).head(30)
```

```python
# ============================================================
# BUILD HISTORICAL SNAPSHOT
# ============================================================

def create_historical_snapshot(data, cutoff_year):
    """
    Returns only matches that occurred on or before the
    specified cutoff year.

    This prevents future information from leaking into
    historical Powerhouse Index calculations.
    """

    snapshot = data[
        data["year"] <= cutoff_year
    ].copy()

    return snapshot


# Test the function
for wc_year, cutoff_year in backtest_world_cups.items():

    snapshot = create_historical_snapshot(
        df2,
        cutoff_year
    )

    print(
        f"{wc_year} World Cup | "
        f"Cutoff: {cutoff_year} | "
        f"Matches available: {len(snapshot):,} | "
        f"Latest year: {snapshot['year'].max()}"
    )
```

```python
def build_powerhouse_index_as_of(cutoff_year):
    """
    Rebuild the Powerhouse Index using only information
    available through cutoff_year.
    """

    # --------------------------------------------------------
    # 1. Historical match snapshot
    # --------------------------------------------------------
    snapshot = df2[df2["year"] <= cutoff_year].copy()

    # --------------------------------------------------------
    # 2. Elo history through cutoff
    # --------------------------------------------------------
    _, snapshot_elo_history = run_elo_with_history(snapshot)

    snapshot_elo_history["year"] = (
        snapshot_elo_history["date"].dt.year
    )

    year_end = (
        snapshot_elo_history
        .sort_values("date")
        .groupby(["team", "year"])
        .tail(1)
        [["team", "year", "elo"]]
    )

    # Use post-1950 global era
    global_elo = year_end[
        year_end["year"] >= 1950
    ].copy()

    # Germany lineage
    global_elo["powerhouse_team"] = (
        global_elo["team"]
        .replace({
            "West Germany": "Germany"
        })
    )

    # One rating per lineage per year
    global_yearly = (
        global_elo
        .sort_values(["year", "elo"])
        .groupby(["powerhouse_team", "year"])
        .tail(1)
        .copy()
    )

    # Rank nations within each year
    global_yearly["world_rank"] = (
        global_yearly
        .groupby("year")["elo"]
        .rank(
            method="min",
            ascending=False
        )
    )

    # Historical strength / longevity
    dominance = (
        global_yearly
        .assign(
            top10=lambda x: x["world_rank"] <= 10,
            top5=lambda x: x["world_rank"] <= 5
        )
        .groupby("powerhouse_team")
        .agg(
            years_active=("year", "nunique"),
            top10_years=("top10", "sum"),
            top5_years=("top5", "sum"),
            avg_elo=("elo", "mean"),
            peak_elo=("elo", "max")
        )
    )

    dominance["top10_pct"] = (
        dominance["top10_years"]
        / dominance["years_active"]
        * 100
    )

    dominance["top5_pct"] = (
        dominance["top5_years"]
        / dominance["years_active"]
        * 100
    )

    # --------------------------------------------------------
    # 3. World Cup pedigree through cutoff
    # --------------------------------------------------------

    wc_cutoff = wc_pedigree.copy()

    # Rebuild WC achievements from editions completed
    # by the cutoff year
    valid_wc_years = (
        wc_champions[
            wc_champions["year"] <= cutoff_year
        ]["year"]
        .unique()
    )

    wc_titles_cut = (
        wc_champions[
            wc_champions["year"].isin(valid_wc_years)
        ]["champion"]
        .value_counts()
    )

    wc_runnerups_cut = (
        wc_champions[
            wc_champions["year"].isin(valid_wc_years)
        ]["runner_up"]
        .value_counts()
    )

    sf_cut = all_other_sf[
        all_other_sf["year"] <= cutoff_year
    ]

    sf_counts_cut = (
        sf_cut["team"]
        .value_counts()
    )

    wc_cutoff = pd.DataFrame({
        "titles": wc_titles_cut,
        "runner_ups": wc_runnerups_cut,
        "other_semifinals": sf_counts_cut
    }).fillna(0)

    wc_cutoff = wc_cutoff.astype(int)

    wc_cutoff["top4_finishes"] = (
        wc_cutoff["titles"]
        + wc_cutoff["runner_ups"]
        + wc_cutoff["other_semifinals"]
    )

    wc_cutoff["wc_score"] = (
        10 * wc_cutoff["titles"]
        + 6 * wc_cutoff["runner_ups"]
        + 2 * wc_cutoff["other_semifinals"]
    )

    # --------------------------------------------------------
    # 4. Continental titles through cutoff
    # --------------------------------------------------------

    champions_cut = champions_df[
        champions_df["year"] <= cutoff_year
    ].copy()

    continental_titles_cut = (
        champions_cut
        .groupby("champion")
        .agg(
            continental_titles=("champion", "size"),
            weighted_title_score=("title_weight", "sum")
        )
    )

    # --------------------------------------------------------
    # 5. Continental dominance through cutoff
    # --------------------------------------------------------

    continental_cut = continental[
        continental["year"] <= cutoff_year
    ].copy()

    continental_cut["home_points"] = 0
    continental_cut["away_points"] = 0

    home_win = (
        continental_cut["home_score"]
        > continental_cut["away_score"]
    )

    away_win = (
        continental_cut["away_score"]
        > continental_cut["home_score"]
    )

    draw = (
        continental_cut["home_score"]
        == continental_cut["away_score"]
    )

    continental_cut.loc[
        home_win, "home_points"
    ] = 3

    continental_cut.loc[
        away_win, "away_points"
    ] = 3

    continental_cut.loc[
        draw,
        ["home_points", "away_points"]
    ] = 1

    cont_home = pd.DataFrame({
        "team": continental_cut["powerhouse_home"],
        "competition": continental_cut["competition"],
        "GF": continental_cut["home_score"],
        "GA": continental_cut["away_score"],
        "points": continental_cut["home_points"]
    })

    cont_away = pd.DataFrame({
        "team": continental_cut["powerhouse_away"],
        "competition": continental_cut["competition"],
        "GF": continental_cut["away_score"],
        "GA": continental_cut["home_score"],
        "points": continental_cut["away_points"]
    })

    cont_team_matches = pd.concat(
        [cont_home, cont_away],
        ignore_index=True
    )

    cont_team_matches["GD"] = (
        cont_team_matches["GF"]
        - cont_team_matches["GA"]
    )

    cont_team_matches["win"] = (
        cont_team_matches["points"] == 3
    ).astype(int)

    cont_perf = (
        cont_team_matches
        .groupby("team")
        .agg(
            matches=("team", "size"),
            wins=("win", "sum"),
            goal_difference=("GD", "sum"),
            points=("points", "sum")
        )
    )

    cont_perf["win_pct"] = (
        cont_perf["wins"]
        / cont_perf["matches"]
        * 100
    )

    cont_perf["points_per_game"] = (
        cont_perf["points"]
        / cont_perf["matches"]
    )

    cont_perf["goal_difference_per_game"] = (
        cont_perf["goal_difference"]
        / cont_perf["matches"]
    )

    team_comp = (
        cont_team_matches
        .groupby("team")["competition"]
        .agg(
            lambda x: x.value_counts().index[0]
        )
    )

    cont_perf["competition"] = team_comp

    cont_valid = cont_perf[
        cont_perf["matches"] >= 20
    ].copy()

    cont_valid["ppg_z"] = (
        cont_valid
        .groupby("competition")["points_per_game"]
        .transform(
            lambda x: (x - x.mean()) / x.std()
        )
    )

    cont_valid["gd_z"] = (
        cont_valid
        .groupby("competition")[
            "goal_difference_per_game"
        ]
        .transform(
            lambda x: (x - x.mean()) / x.std()
        )
    )

    cont_valid["win_z"] = (
        cont_valid
        .groupby("competition")["win_pct"]
        .transform(
            lambda x: (x - x.mean()) / x.std()
        )
    )

    cont_valid["continental_dominance"] = (
        0.50 * cont_valid["ppg_z"]
        + 0.30 * cont_valid["gd_z"]
        + 0.20 * cont_valid["win_z"]
    )

    cont_valid["tournament_weight"] = (
        cont_valid["competition"]
        .map(tournament_weights)
    )

    cont_valid[
        "weighted_continental_dominance"
    ] = (
        cont_valid["continental_dominance"]
        * cont_valid["tournament_weight"]
    )

    continental_cut_component = (
        cont_valid[
            [
                "competition",
                "matches",
                "weighted_continental_dominance"
            ]
        ]
        .join(
            continental_titles_cut,
            how="outer"
        )
    )

    # --------------------------------------------------------
    # 6. Merge historical components
    # --------------------------------------------------------

    master = dominance.copy()

    master = master.join(
        wc_cutoff[
            [
                "titles",
                "runner_ups",
                "other_semifinals",
                "top4_finishes",
                "wc_score"
            ]
        ],
        how="left"
    )

    master = master.join(
        continental_cut_component[
            [
                "continental_titles",
                "weighted_title_score",
                "weighted_continental_dominance"
            ]
        ],
        how="left"
    )

    fill_zero = [
        "titles",
        "runner_ups",
        "other_semifinals",
        "top4_finishes",
        "wc_score",
        "continental_titles",
        "weighted_title_score",
        "weighted_continental_dominance"
    ]

    master[fill_zero] = (
        master[fill_zero]
        .fillna(0)
    )

    # --------------------------------------------------------
    # 7. Normalize same features as final model
    # --------------------------------------------------------

    score_features = [
        "wc_score",
        "top5_years",
        "avg_elo",
        "peak_elo",
        "weighted_title_score",
        "weighted_continental_dominance"
    ]

    scaler = MinMaxScaler(
        feature_range=(0, 100)
    )

    normalized = pd.DataFrame(
        scaler.fit_transform(
            master[score_features]
        ),
        columns=[
            f"{x}_norm"
            for x in score_features
        ],
        index=master.index
    )

    master = master.join(normalized)

    # --------------------------------------------------------
    # 8. Same four component scores
    # --------------------------------------------------------

    master["world_stage_score"] = (
        master["wc_score_norm"]
    )

    master["sustained_strength_score"] = (
        0.50 * master["top5_years_norm"]
        + 0.50 * master["avg_elo_norm"]
    )

    master["continental_score"] = (
        0.50 * master["weighted_title_score_norm"]
        + 0.50 *
        master[
            "weighted_continental_dominance_norm"
        ]
    )

    master["peak_strength_score"] = (
        master["peak_elo_norm"]
    )

    # --------------------------------------------------------
    # 9. SAME FINAL POWERHOUSE WEIGHTS
    # --------------------------------------------------------

    master["powerhouse_index"] = (
        0.35 * master["world_stage_score"]
        + 0.30 *
        master["sustained_strength_score"]
        + 0.20 *
        master["continental_score"]
        + 0.15 *
        master["peak_strength_score"]
    )

    rankings = (
        master
        .sort_values(
            "powerhouse_index",
            ascending=False
        )
        .copy()
    )

    rankings["rank"] = (
        rankings["powerhouse_index"]
        .rank(
            ascending=False,
            method="min"
        )
        .astype(int)
    )

    return rankings
```

```python
# ============================================================
# WORLD CUP SEMIFINALIST BACKTEST
# ============================================================

backtest_years = [2010, 2014, 2018, 2022, 2026]

semifinal_backtest = []

for wc_year in backtest_years:

    cutoff_year = wc_year - 1

    # Build rankings using ONLY information available
    # before the World Cup
    rankings = build_powerhouse_index_as_of(cutoff_year)

    # Get World Cup matches for this tournament
    tournament = df2[
        (df2["tournament"] == "World Cup") &
        (df2["year"] == wc_year)
    ].copy()

    # Sort chronologically
    tournament = tournament.sort_values("date")

    # Last 4 matches should contain:
    # 2 semifinals + third-place match + final
    final_four_matches = tournament.tail(4)

    semifinalists = set(
        final_four_matches["home_team"]
    ) | set(
        final_four_matches["away_team"]
    )

    for team in semifinalists:

        if team in rankings.index:

            semifinal_backtest.append({
                "world_cup": wc_year,
                "cutoff_year": cutoff_year,
                "team": team,
                "pre_wc_rank": rankings.loc[team, "rank"],
                "powerhouse_index":
                    rankings.loc[team, "powerhouse_index"]
            })


semifinal_backtest = pd.DataFrame(semifinal_backtest)

semifinal_backtest.sort_values(
    ["world_cup", "pre_wc_rank"]
)
```
